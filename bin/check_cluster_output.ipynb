## GraphST

```python
# Load libraries
import scanpy as sc
import anndata
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data
load_path = "../data/tools/graphst/"
atac_file = "sp_atac_graphst.h5ad"
sp_file = "sp_graphst.h5ad"

atac_ad = sc.read_h5ad(os.path.join(load_path, atac_file))
sp_ad = sc.read_h5ad(os.path.join(load_path, sp_file))
```

```python
atac_ad.obs
```

```python
sp_ad.obs
```

I have the same cell ids, so I can use them to compare the clustering results

```python
# Compare clustering (mclust column) based on cell ids
# I have the same cell ids, so I can use them to compare the clustering results

# First confirm that the cell ids match
assert all(atac_ad.obs_names == sp_ad.obs_names), "Cell IDs do not match between datasets."
```

```python
# Now get the clustering results
atac_clusters = atac_ad.obs['mclust']
sp_clusters = sp_ad.obs['mclust']
```

```python
# Count the number of matching cluster assignments
matching_clusters = sum(atac_clusters == sp_clusters)
total_cells = len(atac_clusters)
print(f"Number of matching cluster assignments: {matching_clusters} ({matching_clusters / total_cells:.2%}) out of {total_cells} cells.")
```

```python
# Visualize the comparison with a confusion matrix
# As cluster sizes differ greatly, use row-normalized (relative) frequencies

confusion_matrix = pd.crosstab(atac_clusters, sp_clusters, rownames=['ATAC Clusters'], colnames=['SP Clusters'], normalize='index')
plt.figure(figsize=(10, 8))
sns.heatmap(confusion_matrix, annot=True, fmt='.2f', cmap='Blues')
plt.gca().invert_yaxis() 
plt.gca().set_title('Confusion Matrix of Cluster Assignments', y=1.03)
plt.subplots_adjust(top=0.92)
plt.show()
```

## StLearn

```python
# Load libraries
import scanpy as sc
import anndata
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load StLearn results
load_path = "../data/tools/stlearn/"
atac_file = "sp_atac_stlearn.h5ad"
sp_file = "sp_stlearn.h5ad"
atac_ad = sc.read_h5ad(os.path.join(load_path, atac_file))
sp_ad = sc.read_h5ad(os.path.join(load_path, sp_file))
```

```python
sp_ad.obs
```

```python
atac_ad.obs
```

```python
assert all(atac_ad.obs_names == sp_ad.obs_names), "Cell IDs do not match between datasets."
```

```python
# Get the clustering results
atac_clusters_kmeans = atac_ad.obs['stlearn_kmeans']
sp_clusters_kmeans = sp_ad.obs['stlearn_kmeans']

atac_clusters_louvain   = atac_ad.obs['stlearn_louvain']
sp_clusters_louvain = sp_ad.obs['stlearn_louvain']
```

```python
# Check matching cluster assignments
matching_clusters = sum(atac_clusters_louvain.astype(str).values == sp_clusters_louvain.astype(str).values)
total_cells = len(atac_clusters_louvain)
print(f"Number of matching cluster assignments: {matching_clusters} ({matching_clusters / total_cells:.2%}) out of {total_cells} cells.")
```

```python
# Same for kmeans
matching_clusters_kmeans = sum(atac_clusters_kmeans.astype(str).values == sp_clusters_kmeans.astype(str).values)
total_cells = len(atac_clusters_kmeans)
print(f"Number of matching kmeans cluster assignments: {matching_clusters_kmeans} ({matching_clusters_kmeans / total_cells:.2%}) out of {total_cells} cells.")
```

```python
# Plot confusion matrix
confusion_matrix = pd.crosstab(atac_clusters_kmeans, sp_clusters_kmeans, rownames=['ATAC Clusters'], colnames=['SP Clusters'], normalize='index')
plt.figure(figsize=(10, 8))
sns.heatmap(confusion_matrix, annot=True, fmt='.2f', cmap='Blues')
plt.gca().invert_yaxis() 
plt.gca().set_title('Confusion Matrix of Cluster Assignments (KMeans)', y=1.03)
plt.subplots_adjust(top=0.92)
plt.show()
```

```python
# Plot confusion matrix
confusion_matrix = pd.crosstab(atac_clusters_louvain, sp_clusters_louvain, rownames=['ATAC Clusters'], colnames=['SP Clusters'], normalize='index')
plt.figure(figsize=(10, 8))
sns.heatmap(confusion_matrix, annot=True, fmt='.2f', cmap='Blues')
plt.gca().invert_yaxis() 
plt.gca().set_title('Confusion Matrix of Cluster Assignments (Louvain)', y=1.03)
plt.subplots_adjust(top=0.92)
plt.show()
```
