## GraphST

```python
# Load libraries
import scanpy as sc
import anndata
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from scipy.optimize import linear_sum_assignment

def match_clusters(true_labels, pred_labels):
    """
    Matches clusters between two labelings using the Hungarian algorithm.
    Returns a mapping from pred_labels to true_labels.
    """
    true_labels = pd.Series(true_labels).astype(str)
    pred_labels = pd.Series(pred_labels).astype(str)
    true_cats = true_labels.unique()
    pred_cats = pred_labels.unique()
    # Always match to the smaller set
    if len(true_cats) <= len(pred_cats):
        row_labels, col_labels = true_cats, pred_cats
        row = 'true'
    else:
        row_labels, col_labels = pred_cats, true_cats
        row = 'pred'
    # Build confusion matrix
    cm = pd.crosstab(true_labels, pred_labels)
    cm = cm.reindex(index=row_labels, columns=col_labels, fill_value=0)
    # Convert to cost matrix (maximize overlap)
    cost_matrix = -cm.values
    row_ind, col_ind = linear_sum_assignment(cost_matrix)
    mapping = {}
    for i, j in zip(row_ind, col_ind):
        if row == 'true':
            mapping[col_labels[j]] = row_labels[i]
        else:
            mapping[row_labels[i]] = col_labels[j]
    return mapping

# Example usage:
# mapping = match_clusters(atac_clusters, sp_clusters)
# mapped_sp_clusters = sp_clusters.map(mapping)
```

```python
# Load data
load_path = "../data/tools/graphst/"
atac_file = "sp_atac_graphst.h5ad"
sp_file = "sp_graphst.h5ad"

atac_ad = sc.read_h5ad(os.path.join(load_path, atac_file))
sp_ad = sc.read_h5ad(os.path.join(load_path, sp_file))
```

```python
atac_ad.obs
```

```python
sp_ad.obs
```

I have the same cell ids, so I can use them to compare the clustering results

```python
# Compare clustering (mclust column) based on cell ids
# I have the same cell ids, so I can use them to compare the clustering results

# First confirm that the cell ids match
assert all(atac_ad.obs_names == sp_ad.obs_names), "Cell IDs do not match between datasets."
```

```python
# Now get the clustering results
atac_clusters = atac_ad.obs['mclust']
sp_clusters = sp_ad.obs['mclust']
```

```python
atac_clusters
```

```python
# Count the number of matching cluster assignments
matching_clusters = sum(atac_clusters == sp_clusters)
total_cells = len(atac_clusters)
print(f"Number of matching cluster assignments: {matching_clusters} ({matching_clusters / total_cells:.2%}) out of {total_cells} cells.")
```

```python
mapping = match_clusters(atac_clusters, sp_clusters)
```

```python
mapping
```

```python
# Visualize the comparison with a confusion matrix
# As cluster sizes differ greatly, use row-normalized (relative) frequencies

confusion_matrix = pd.crosstab(atac_clusters, sp_clusters, rownames=['ATAC Clusters'], colnames=['SP Clusters'], normalize='index')
plt.figure(figsize=(10, 8))
sns.heatmap(confusion_matrix, annot=True, fmt='.2f', cmap='Blues')
plt.gca().invert_yaxis() 
plt.gca().set_title('Confusion Matrix of Cluster Assignments', y=1.03)
plt.subplots_adjust(top=0.92)
plt.show()
```

```python
# Visualize the comparison with a confusion matrix
confusion_matrix = pd.crosstab(atac_clusters, sp_clusters, rownames=['ATAC Clusters'], colnames=['SP Clusters'], normalize='index')
# Get the matching pairs from the mapping
pairs = sorted(mapping.items(), key=lambda x: x[1])

# Extract the order for rows and columns
ordered_rows = [int(v) for k, v in pairs]
ordered_cols = [int(k) for k, v in pairs]

# Reorder both axes
confusion_matrix = confusion_matrix.loc[ordered_rows, ordered_cols]

plt.figure(figsize=(10, 8))
sns.heatmap(confusion_matrix, annot=True, fmt='.2f', cmap='Blues')
plt.gca().invert_yaxis()
plt.gca().set_title('Confusion Matrix of Cluster Assignments', y=1.03)
plt.subplots_adjust(top=0.92)
plt.show()
```
